```python
import deeplabcut

video = "data/cam3.mp4"   # any one camera video
video_name = video.split('/')[-1]

deeplabcut.video_inference_superanimal(
    [video],
    superanimal_name="full_cheetah",
    model_name="hrnet_w32",                       # top-down pose model
    detector_name="fasterrcnn_resnet50_fpn_v2",   # finds the cheetah first
    max_individuals=1,                            # one cheetah per video
    device="cuda",
    dest_folder=f"results/{video_name}",
    # bbox_threshold=0.5
)
```
